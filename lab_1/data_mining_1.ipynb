## Задание 1

```python
import requests
import json

username = "RinatHar"
url = f"https://api.github.com/users/{username}/repos"

response = requests.get(url)

if response.status_code == 200:
    repos = response.json()
    print(f"Найдено репозиториев: {len(repos)}")

    with open(f"{username}_repos.json", "w", encoding="utf-8") as f:
        json.dump(repos, f, ensure_ascii=False, indent=4)

    print(f"Файл {username}_repos.json сохранён")

    print("\nСписок репозиториев:")
    for repo in repos:
        print(" -", repo["name"])
else:
    print("Ошибка запроса:", response.text)
```

## Задание 2

Выбран NewsAPI - https://newsapi.org

```python
import requests
import json
from getpass import getpass

# Вводим ключ вручную, чтобы не хранить его в блокноте
api_key = getpass("Введите ваш NewsAPI ключ: ")

query = "цены на продукты"  # ключевое слово
language = "ru"             # язык
sort_by = "relevancy"       # сортировка: relevancy | popularity | publishedAt
page_size = 20              # сколько статей вернуть

url = "https://newsapi.org/v2/everything"
params = {
    "q": query,
    "language": language,
    "sortBy": sort_by,
    "pageSize": page_size,
}
headers = {
    "X-Api-Key": api_key,
}

response = requests.get(url, params=params, headers=headers)
data = response.json()

if response.status_code != 200 or data.get("status") != "ok":
    print("Ошибка API:", data.get("message", "неизвестная ошибка"))
else:
    print("Статус-код:", response.status_code)
    print("Всего найдено статей:", data["totalResults"])
    print(f"Получено в ответе: {len(data['articles'])}\n")

    # Краткий вывод заголовков
    for i, article in enumerate(data["articles"], 1):
        print(f"{i}. {article['title']}")
        print(f"   Источник: {article['source']['name']}")
        print(f"   Дата: {article['publishedAt']}")
        print(f"   URL: {article['url']}\n")

    # Сохраняем полный JSON-ответ в файл
    with open("news.json", "w", encoding="utf-8") as f:
        json.dump(data, f, ensure_ascii=False, indent=4)

    print("Файл news.json сохранён")
```
